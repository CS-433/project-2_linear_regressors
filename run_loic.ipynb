```python
import pandas as pd
from datasets import Dataset
from evaluate import load
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from transformers import TrainingArguments, Trainer
from sklearn.model_selection import train_test_split
import torch

if torch.backends.mps.is_available():
    device = torch.device("mps")
    print(">>> Using Apple MPS GPU")
else:
    device = torch.device("cpu")
    print(">>> MPS not available, using CPU")

# ============================================================
# 1) LOAD DATA
# ============================================================

with open("twitter-datasets/train_pos_full.txt", "r") as f:
    pos = [line.strip() for line in f]

with open("twitter-datasets/train_neg_full.txt", "r") as f:
    neg = [line.strip() for line in f]

train_texts = pos + neg
train_labels = [1] * len(pos) + [0] * len(neg)  # 1=pos, 0=neg

train_df = pd.DataFrame({"text": train_texts, "label": train_labels})

# -------- split into train / valid --------
train_df, valid_df = train_test_split(
    train_df, test_size=0.1, random_state=42, shuffle=True
)

# ================================
# REDUCE DATASET SIZE (FAST TRAIN)
# ================================

# Shuffle globally (just in case)
train_df = train_df.sample(frac=1, random_state=42)

# Sample 100k for training
train_df = train_df.sample(100_000, random_state=42)

# Sample 10k for validation
valid_df = valid_df.sample(10_000, random_state=42)

print("Train size:", len(train_df))
print("Valid size:", len(valid_df))

# Load test set
test_texts = []
with open("twitter-datasets/test_data.txt", "r") as f:
    for line in f:
        line = line.strip()
        tweet = line.split(",", 1)[1]
        test_texts.append(tweet)

test_df = pd.DataFrame({"text": test_texts})

# Convert to HF datasets
train_dataset = Dataset.from_pandas(train_df)
valid_dataset = Dataset.from_pandas(valid_df)
test_dataset = Dataset.from_pandas(test_df)

# ============================================================
# 2) TOKENIZATION
# ============================================================

model_name = "vinai/bertweet-base"
tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)

def tokenize_fn(batch):
    return tokenizer(
        batch["text"],
        padding="max_length",
        truncation=True,
        max_length=64
    )

train_dataset = train_dataset.map(tokenize_fn, batched=True)
valid_dataset = valid_dataset.map(tokenize_fn, batched=True)
test_dataset = test_dataset.map(tokenize_fn, batched=True)

train_dataset = train_dataset.remove_columns(["text"])
valid_dataset = valid_dataset.remove_columns(["text"])
test_dataset = test_dataset.remove_columns(["text"])

train_dataset = train_dataset.with_format("torch")
valid_dataset = valid_dataset.with_format("torch")
test_dataset = test_dataset.with_format("torch")

# ============================================================
# 3) MODEL
# ============================================================

model = AutoModelForSequenceClassification.from_pretrained(
    model_name, num_labels=2
).to(device)

# ============================================================
# 4) METRIC
# ============================================================

accuracy = load("accuracy")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = logits.argmax(axis=-1)
    return accuracy.compute(predictions=preds, references=labels)

# ============================================================
# 5) TRAINING ARGUMENTS
# ============================================================

training_args = TrainingArguments(
    output_dir="bertweet_sentiment",
    evaluation_strategy="epoch",
    save_strategy="no",
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    learning_rate=2e-5,
    num_train_epochs=2,
    weight_decay=0.01,
    logging_steps=200,
    report_to="none",
    no_cuda=False if device.type == "mps" else None,
    fp16=False if device.type == "mps" else None,
    bf16=True if device.type == "mps" else None,

)

# ============================================================
# 6) TRAINER
# ============================================================

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=valid_dataset,
    compute_metrics=compute_metrics
)

trainer.model.to(device)

# ============================================================
# 7) TRAIN
# ============================================================

trainer.train()

# ============================================================
# 8) PREDICT TEST SET
# ============================================================

preds = trainer.predict(test_dataset)
pred_logits = torch.tensor(preds.predictions, device=device)
prob = torch.softmax(pred_logits, dim=1)
labels = torch.argmax(prob, dim=1).cpu().numpy()

final_preds = [1 if x == 1 else -1 for x in labels]

# ============================================================
# 9) METRICS ON VALID SET
# ============================================================

import numpy as np
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix
)
import matplotlib.pyplot as plt
import seaborn as sns

print("\n>>> Evaluating on validation set...")

# Recompute predictions on validation set
valid_preds = trainer.predict(valid_dataset)
valid_logits = valid_preds.predictions
valid_pred_labels = np.argmax(valid_logits, axis=1)
valid_true_labels = valid_dataset["label"]

# ---- Metrics ----
acc = accuracy_score(valid_true_labels, valid_pred_labels)
prec = precision_score(valid_true_labels, valid_pred_labels, zero_division=0)
rec = recall_score(valid_true_labels, valid_pred_labels, zero_division=0)
f1 = f1_score(valid_true_labels, valid_pred_labels, zero_division=0)

print("\n===== VALIDATION METRICS =====")
print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1-score : {f1:.4f}")

# ---- Confusion matrix ----
cm = confusion_matrix(valid_true_labels, valid_pred_labels)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["neg", "pos"],
    yticklabels=["neg", "pos"]
)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix (Validation)")
plt.show()


# ============================================================
# 10) SAVE SUBMISSION FILE
# ============================================================

import pandas as pd

submission = pd.DataFrame({
    "Id": np.arange(1, len(final_preds) + 1),
    "Prediction": final_preds
})

submission.to_csv("submit.csv", index=False)
print(">>> Saved submit.csv")


# ============================================================
# 11) SAVE MODEL + TOKENIZER
# ============================================================

save_dir = "bertweet_finetuned_100k"

print(f">>> Saving model to {save_dir}/")

model.to("cpu")   # safer for save
model.save_pretrained(save_dir)
tokenizer.save_pretrained(save_dir)

print(">>> Model + tokenizer saved successfully!")

```

```python
import os
import pandas as pd
from datasets import Dataset
from evaluate import load
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from transformers import TrainingArguments, Trainer, TrainerCallback
from sklearn.model_selection import train_test_split
import torch
import numpy as np
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix
)
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# CUSTOM LOGGING CALLBACK
# ============================================================

class MetricsLogger(TrainerCallback):
    def __init__(self):
        self.train_loss = []
        self.eval_loss = []
        self.eval_acc = []

    def on_log(self, args, state, control, logs=None, **kwargs):
        if logs is None:
            return

        # Training loss
        if "loss" in logs:
            self.train_loss.append((state.global_step, logs["loss"]))

        # Eval loss
        if "eval_loss" in logs:
            self.eval_loss.append((state.global_step, logs["eval_loss"]))

        # Eval accuracy
        if "eval_accuracy" in logs:
            self.eval_acc.append((state.global_step, logs["eval_accuracy"]))


# ============================================================
# MAIN FUNCTION
# ============================================================

def run_training(TRAIN_SIZE, VALID_SIZE, model_name):

    # Directory
    RESULTS_DIR = "results"
    os.makedirs(RESULTS_DIR, exist_ok=True)

    # tags
    model_tag = model_name.split("/")[-1]
    run_tag = f"{model_tag}_train{TRAIN_SIZE}"

    # ============================================================
    # DEVICE
    # ============================================================

    if torch.backends.mps.is_available():
        device = torch.device("mps")
        print(">>> Using Apple MPS GPU")
    else:
        device = torch.device("cpu")
        print(">>> MPS not available, using CPU")

    # ============================================================
    # 1) LOAD DATA
    # ============================================================

    with open("twitter-datasets/train_pos_full.txt", "r") as f:
        pos = [line.strip() for line in f]

    with open("twitter-datasets/train_neg_full.txt", "r") as f:
        neg = [line.strip() for line in f]

    train_texts = pos + neg
    train_labels = [1] * len(pos) + [0] * len(neg)

    train_df = pd.DataFrame({"text": train_texts, "label": train_labels})

    train_df, valid_df = train_test_split(
        train_df, test_size=0.1, random_state=42, shuffle=True
    )

    train_df = train_df.sample(frac=1, random_state=42)

    train_df = train_df.sample(TRAIN_SIZE, random_state=42)
    valid_df = valid_df.sample(VALID_SIZE, random_state=42)

    print("Train size:", len(train_df))
    print("Valid size:", len(valid_df))

    # load test set
    test_texts = []
    with open("twitter-datasets/test_data.txt", "r") as f:
        for line in f:
            line = line.strip()
            tweet = line.split(",", 1)[1]
            test_texts.append(tweet)

    test_df = pd.DataFrame({"text": test_texts})

    train_dataset = Dataset.from_pandas(train_df)
    valid_dataset = Dataset.from_pandas(valid_df)
    test_dataset = Dataset.from_pandas(test_df)

    # ============================================================
    # 2) TOKENIZATION
    # ============================================================

    tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)

    def tokenize_fn(batch):
        return tokenizer(
            batch["text"],
            padding="max_length",
            truncation=True,
            max_length=64
        )

    train_dataset = train_dataset.map(tokenize_fn, batched=True)
    valid_dataset = valid_dataset.map(tokenize_fn, batched=True)
    test_dataset = test_dataset.map(tokenize_fn, batched=True)

    train_dataset = train_dataset.remove_columns(["text"])
    valid_dataset = valid_dataset.remove_columns(["text"])
    test_dataset = test_dataset.remove_columns(["text"])

    train_dataset = train_dataset.with_format("torch")
    valid_dataset = valid_dataset.with_format("torch")
    test_dataset = test_dataset.with_format("torch")

    # ============================================================
    # 3) MODEL
    # ============================================================

    model = AutoModelForSequenceClassification.from_pretrained(
        model_name, num_labels=2
    ).to(device)

    # ============================================================
    # 4) METRIC
    # ============================================================

    accuracy = load("accuracy")

    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        preds = logits.argmax(axis=-1)
        return accuracy.compute(predictions=preds, references=labels)

    # ============================================================
    # 5) TRAINING ARGUMENTS
    # ============================================================

    training_args = TrainingArguments(
        output_dir=f"bertweet_sentiment_{run_tag}",
        evaluation_strategy="epoch",
        save_strategy="no",
        per_device_train_batch_size=16,
        per_device_eval_batch_size=32,
        learning_rate=2e-5,
        num_train_epochs=2,
        weight_decay=0.01,
        logging_steps=200,
        report_to="none",
        no_cuda=False if device.type == "mps" else None,
        fp16=False if device.type == "mps" else None,
        bf16=True if device.type == "mps" else None,
    )

    # ============================================================
    # 6) TRAINER
    # ============================================================

    logger = MetricsLogger()

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=valid_dataset,
        compute_metrics=compute_metrics,
        callbacks=[logger]
    )

    trainer.model.to(device)

    # ============================================================
    # 7) TRAIN
    # ============================================================

    trainer.train()

    # ============================================================
    # 8) PREDICT TEST SET
    # ============================================================

    preds = trainer.predict(test_dataset)
    pred_logits = torch.tensor(preds.predictions, device=device)
    prob = torch.softmax(pred_logits, dim=1)
    labels = torch.argmax(prob, dim=1).cpu().numpy()

    final_preds = [1 if x == 1 else -1 for x in labels]

    # ============================================================
    # 9) METRICS ON VALID SET
    # ============================================================

    valid_preds = trainer.predict(valid_dataset)
    valid_logits = valid_preds.predictions
    valid_pred_labels = np.argmax(valid_logits, axis=1)
    valid_true_labels = valid_dataset["label"]

    acc = accuracy_score(valid_true_labels, valid_pred_labels)
    prec = precision_score(valid_true_labels, valid_pred_labels, zero_division=0)
    rec = recall_score(valid_true_labels, valid_pred_labels, zero_division=0)
    f1 = f1_score(valid_true_labels, valid_pred_labels, zero_division=0)

    cm = confusion_matrix(valid_true_labels, valid_pred_labels)

    # ============================================================
    # SAVE CONFUSION MATRIX
    # ============================================================

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["neg", "pos"],
        yticklabels=["neg", "pos"]
    )
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(f"Confusion Matrix – {model_tag} – train={TRAIN_SIZE}")

    cm_path = os.path.join(RESULTS_DIR, f"confusion_{run_tag}.png")
    plt.savefig(cm_path, dpi=200, bbox_inches="tight")
    plt.close()

    print(f">>> Saved {cm_path}")

    # ============================================================
    # 10) SAVE TRAINING CURVES
    # ============================================================

    # ------ TRAIN LOSS ------
    steps, losses = zip(*logger.train_loss)
    plt.plot(steps, losses)
    plt.title(f"Train Loss – {model_tag} – train={TRAIN_SIZE}")
    plt.xlabel("Steps")
    plt.ylabel("Loss")
    path = os.path.join(RESULTS_DIR, f"train_loss_{run_tag}.png")
    plt.savefig(path, dpi=200, bbox_inches="tight")
    plt.close()
    print(f">>> Saved {path}")

    # ------ EVAL LOSS ------
    if logger.eval_loss:
        steps, losses = zip(*logger.eval_loss)
        plt.plot(steps, losses)
        plt.title(f"Eval Loss – {model_tag} – train={TRAIN_SIZE}")
        plt.xlabel("Steps")
        plt.ylabel("Loss")
        path = os.path.join(RESULTS_DIR, f"eval_loss_{run_tag}.png")
        plt.savefig(path, dpi=200, bbox_inches="tight")
        plt.close()
        print(f">>> Saved {path}")

    # ------ EVAL ACC ------
    if logger.eval_acc:
        steps, accs = zip(*logger.eval_acc)
        plt.plot(steps, accs)
        plt.title(f"Eval Accuracy – {model_tag} – train={TRAIN_SIZE}")
        plt.xlabel("Steps")
        plt.ylabel("Accuracy")
        path = os.path.join(RESULTS_DIR, f"eval_acc_{run_tag}.png")
        plt.savefig(path, dpi=200, bbox_inches="tight")
        plt.close()
        print(f">>> Saved {path}")

    # ============================================================
    # 11) SAVE SUBMISSION
    # ============================================================

    submission = pd.DataFrame({
        "Id": np.arange(1, len(final_preds) + 1),
        "Prediction": final_preds
    })

    csv_path = os.path.join(RESULTS_DIR, f"submit_{run_tag}.csv")
    submission.to_csv(csv_path, index=False)
    print(f">>> Saved {csv_path}")

    # ============================================================
    # 12) SAVE MODEL
    # ============================================================

    save_dir = f"{model_tag}_finetuned_{TRAIN_SIZE}"
    os.makedirs(save_dir, exist_ok=True)

    model.to("cpu")
    model.save_pretrained(save_dir)
    tokenizer.save_pretrained(save_dir)

    print(">>> Model + tokenizer saved successfully!")


# ============================================================
# CALL EXAMPLE
# ============================================================

# run_training(100_000, 10_000, "vinai/bertweet-base")
```
